```python
sample_dataset = "1.I absolutely LOVE this phone!! Battery lasts 2 days.  2.Worst Product Ever :( Money Wasted !!   3.Delivery was quick, but the packaging was damaged :( :(   4.Excellent camera quality and amazing performance !!"

# 1. Convert text to lowercase
sample_dataset = sample_dataset.lower()
print(sample_dataset)

# 2. Remove Punctuation
import string
sample_dataset = sample_dataset.translate(str.maketrans('', '', string.punctuation))
print(sample_dataset)

# 3. Remove Numbers
import re
sample_dataset = re.sub(r'\d+', '', sample_dataset)
print(sample_dataset)

# 4. Remove StopWords
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))
words = word_tokenize(sample_dataset)
filtered_words = [word for word in words if word.lower() not in stop_words]
sample_dataset = ' '.join(filtered_words)
print(sample_dataset)

# 5. Define a sample email
sample_email = "Subject: Urgent: Your account has been compromised! Click this link now: http://malicious.com/phish123 Get FREE money! Limited time offer! Call 1-800-spam-now. "
print(f"Original Email:\n{sample_email}")

# 6. Convert email to lowercase
sample_email = sample_email.lower()
print(f"\nAfter Lowercasing:\n{sample_email}")

# 7. Remove Punctuation from email
sample_email = sample_email.translate(str.maketrans('', '', string.punctuation))
print(f"\nAfter Punctuation Removal:\n{sample_email}")

# 8. Remove Numbers from email
sample_email = re.sub(r'\d+', '', sample_email)
print(f"\nAfter Number Removal:\n{sample_email}")

# 9. Remove StopWords from email
stop_words = set(stopwords.words('english'))
email_words = word_tokenize(sample_email)
filtered_email_words = [word for word in email_words if word.lower() not in stop_words]
sample_email_processed = ' '.join(filtered_email_words)
print(f"\nAfter Stopword Removal:\n{sample_email_processed}")
```
